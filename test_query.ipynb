```python
import aijsondb
base_dir="./db/sakila/dataset/json/"
db_file=base_dir+"sakila.json"
db_schema_file=base_dir+"sakila_schema.json"
aijsondb.init_db(db_file, db_schema_file)
```

```python
query="""
let result=[];
    data.stores.map(store => 
       store.Staff.map(staff=>
           result.push(
           {
               "store":store.Address, 
               "staff": staff.First_Name + " "+staff.Last_Name
           }
           )
       )
    )
"""
res=aijsondb.query_data_javascript(query)
if type(res) is list:
    print(len(res))
res
```
