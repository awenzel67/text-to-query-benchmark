```python
# Synchronous Example
from mistralai.client import Mistral
import os

import yaml
with open('db/sakila/questions/sakila_025.yaml', 'r') as file:
    config = yaml.safe_load(file)
print(config['question'])
question=config['question']

user_question=f"""
I have a test database for video rental.
I created the test question:
{question}

Create a slightly different question yielding the same answer.
"""

with Mistral(
    api_key=os.getenv("MISTRAL_API_KEY", ""),
) as mistral:

    res = mistral.chat.complete(model="mistral-large-latest", messages=[
        {
            "role": "user",
            "content": user_question,
        },
    ], stream=False, response_format={
        "type": "text",
    })

    # Handle response
    print(question)
    print("")
    print(res.choices[0].message.content)
```

```python
base_dir_queries="db/sakila/questions"
test_data=[]
for name in os.listdir(base_dir_queries):
    full_path = os.path.join(base_dir_queries, name)
    if os.path.isfile(full_path) and name.endswith('.yaml'):
        testConfig=None
        with open(full_path, 'r') as file:
            testConfig = yaml.safe_load(file)
        test_data.append(testConfig)
```
