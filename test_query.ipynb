```python
import aijsondb
base_dir="./db/sakila/dataset/json/"
db_file=base_dir+"sakila.json"
db_schema_file=base_dir+"sakila_schema.json"
aijsondb.init_db(db_file, db_schema_file)
```

```python
query="""
    let result={};    
    let names=[]
    data.films.map(film=>{
       film.Actors.map(
          actor => {
               if (! (actor.actorId in result))
               {
                   result[actor.actorId]=actor;
                   names.push({"first_name": actor.first_name, "last_name": actor.last_name});
               }
          })
    }); 
    result=names;
"""
```

```python
aijsondb.query_data_javascript(query)
```
