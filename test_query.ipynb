```python
import aijsondb
base_dir="./db/sakila/dataset/json/"
db_file=base_dir+"sakila.json"
db_schema_file=base_dir+"sakila_schema.json"
aijsondb.init_db(db_file, db_schema_file)
```

```python
query="""
let result={};    
    let names=[]
    data.films.map(film=>{
       film.Actors.map(
          actor => {
               name=`${actor.First_name} ${actor.Last_name}`
               if (! (name in result))
               {
                   result[name]=1;
                   names.push(name);
               }
          })
    }); 
    result=names;  
"""
aijsondb.query_data_javascript(query)
```
