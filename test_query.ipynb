```python
import aijsondb
base_dir="./db/sakila/dataset/json/"
db_file=base_dir+"sakila.json"
db_schema_file=base_dir+"sakila_schema.json"
aijsondb.init_db(db_file, db_schema_file)
```

```python
query="""
 let result={};
      data.customers.map(customer=>{
         customer.Rentals.map(rental=>{
            let date = new Date(rental.Rental_Date);
            let month_year=date.getMonth();
            let full_year=date.getFullYear();
            if (full_year==2024)
            {
                if (!(month_year in result)) {
                result[month_year]={"costumers":0,"payments":0};  
                }
                let initialValue=0.0
                let amount=rental.Payments.reduce((accumulator, currentValue) => accumulator + currentValue.Amount,initialValue)
                result[month_year]["costumers"]=result[month_year]["costumers"]+1;
                result[month_year]["payments"]=result[month_year]["payments"]+amount;
            }
         });
      });
"""
res=aijsondb.query_data_javascript(query)
if type(res) is list:
    print(len(res))
res
```
